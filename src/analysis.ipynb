```python
import kagglehub
import pandas as pd
import os

path = kagglehub.dataset_download("uciml/glass")
file_path = os.path.join(path, "glass.csv")

df = pd.read_csv(file_path)

df.head()
```

```python
df.info()
```

Verificação se há duplicadas, como ocorre: A função duplicated verifica se há duplicação, caso seja True equivale a 1 e caso seja False equivale a 0. E o SUM() é uma função no python que soma tudo

```python
duplicados = df.duplicated().sum()
print("Número de duplicatas:", duplicados)

```

O resultado mostra que tem duas linhas iguais

```python
df = df.drop_duplicates() # remorção da duplicata

```

```python
duplicados = df.duplicated().sum()
print(duplicados)
```
